```python
VALID_PASSWORD = "1234"
START_BALANCE: int = 1000

global balance


def main() -> None:
    balance = START_BALANCE

    attempts = 0
    while attempts < 3:
        password = input("Введите пароль: ")
        if password == VALID_PASSWORD:
            print("Доступ разрешен.")
            break
        else:
            attempts += 1
            print(f"Неверный пароль. У вас осталось {3 - attempts} попыток.")
    else:
        print("Карта заблокирована.")
        return

    print("Добро пожаловать в роснацсберфинминторгбанк!")
    print("1 - Пополнить счет")
    print("2 - Снять деньги")
    print("3 - Проверить баланс")
    print("0 - Выход")

    inp: str = "-1"
    while inp != "0":
        try:
            inp = input("Введите число от 0 до 3: ")
            if inp == "1":
                i = int(input("Введите сумму для пополнения баланса: "))
                if i <= 0:
                    print("Некорректная сумма")
                else:
                    balance += i
                    print(f"Пополнено на {i}, баланс: {balance}")
            elif inp == "2":
                i = int(input("Введите сумму для снятия: "))
                if i <= 0:
                    print("Некорректная сумма")
                elif i > balance:
                    print("Недостаточно средств")
                else:
                    balance -= i
                    print(f"Снято {i}, остаток на балансе {balance}")
            elif inp == "3":
                print("Ваш баланс: ", balance)
            elif inp == "0":
                print("До свидания!")
                break
            else:
                print("Некорректный ввод данных. Попробуйте ещё")
        except ValueError:
            print("Некорректный ввод данных. Попробуйте ещё")


main()

```
