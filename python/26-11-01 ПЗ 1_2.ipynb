```python
# 1.1

radius = float(input("Введите радиус круга: "))
square_side = float(input("Введите сторону квадрата: "))
rect_x, rect_y = map(float, input("Введите строны прямоугольника через пробел: ").split())

# 1.2
import math

circle_s = round(math.pi * radius**2, 3)
square_s = round(square_side**2, 3)
rect_s = round(rect_x * rect_y, 3)
# print(circle_s, square_s, rect_s)

# 1.3
circle_c = 2 * math.pi * radius
square_p = 4 * square_side
rect_p = 2 * (rect_x * rect_y)

# 1.4
c_greater_s = circle_s / square_s
r_diff_s = abs(rect_s - square_s)

# 1.5
radius *= 2
square_side *= 2
rect_x *= 2; rect_y *= 2

circle_s_2 = round(math.pi * radius**2, 3)
square_s_2 = round(square_side**2, 3)
rect_s_2 = round(rect_x * rect_y, 3)
print(f"Площаль круга увеличилась в {circle_s_2 // circle_s},\n\
Площадь квадрата увеличилась в {square_s_2 / square_s},\n\
Площаль прям-ка увеличилась в {rect_s_2 / rect_s}")

circle_c_2 = 2 * math.pi * radius
square_p_2 = 4 * square_side
rect_p_2 = 2 * (rect_x * rect_y)

print(f"Периметр круга увеличился в {circle_c_2 / circle_c},\n\
Периметр квадрата увеличился в {square_p_2 / square_p},\n\
Периметр квадрата увеличился в {rect_p_2 / rect_p}")

# 1.6

print("Можно ли вписать круг в квадрат: ",\
      math.hypot(square_side, square_side) <= radius * 2)
print("Можно ли вписать квадрат в круг: ",\
      radius * 2 <= square_side)
print("Можно ли вписать круг в прям-к: ",\
      math.hypot(rect_x, rect_y) <= radius * 2)

# 1.7
print(f"Круг: S = {circle_s_2}, C = {round(circle_c_2, 3)}\n\
Квадрат: S = {square_s_2}, P = {round(square_p_2, 3)} \n\
Прямоугольник: S = {rect_s_2}, P = {round(rect_p_2 ,3)}")

```
