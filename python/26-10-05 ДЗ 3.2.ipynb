**Задание 1.**

Напишите программу, которая принимает на вход строку из чисел через пробел, разбивает строку на список, преобразовывает каждое число в int.  Затем программа печатает новый список, содержащий квадраты каждого числа из исходного списка.

```python
n: list[int] = input("Введите числа через пробел: ").split()
i = [int(x)**2 for x in n]

print(i)
```

**Задание 3.**

На вход вашей программе подается предложение, в котором могут встречаться повторяющиеся слова.

Ваша задача - создать множество, содержащее только уникальные слова, которые строго короче 5 символов. Затем выведите получившиеся слова в отсортированном порядке по алфавиту, разделенные запятой и пробелом.

Обратите внимание, что все слова могут быть написаны в разном регистре, а также в предложении могут встречаться знаки препинания, такие как запятые, точки, восклицательные и вопросительные знаки.

```python
#Ваш код здесь
i: list[str] = input("Введите слова: ").split()
i = [w.lower() for w in i]
d = []
for z in i:
    if len(z) < 5 and i.count(z) == 1:
        d.append(z)

print(d)
```

**Задание 4.**

Напишите программу, которая принимает слово в качестве входных данных и создает словарь, где ключами являются буквы в слове, а значениями - количество раз, которое каждая буква встречается в слове.

Учтите, что регистр букв (верхний или нижний) не должен влиять на подсчет букв.

```python
#Ваш код здесь
i = input("Введите слово: ")
i = [s.lower() for s in i]

d: dict[str, int] = {}

for s in i:
    d[s] = i.count(s)

print(d)
```

**Задание 5.**

Дан текст: в первой строке задано число строк, далее идут сами строки. Выведите слово, которое в этом тексте встречается чаще всего. Если таких слов несколько, выведите то, которое меньше в лексикографическом порядке.

```python
#Ваш код здесь
import string

text = "5 " \
       "О, сколько нам открытий чудных" \
       "Готовят просвещенья дух, " \
       "И гений, сын ошибок трудных, " \
       "И гений, парадокфов друг, " \
       "И гений, бог изобретатель. "

trans = str.maketrans("", "", string.punctuation)

i = text.split()
i = [s.lower().translate(trans) for s in i]

d: dict[str, int] = {}

for s in i:
    d[s] = i.count(s)

keys = [k for k, v in d.items() if v == max(d.values())]

print(min(keys, key=len))
```
