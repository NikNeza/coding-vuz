```python
# Задание 1.1

name = input("Введите имя: ")
math_mark = int(input("Оценка по математике от 1 до 5: "))
phy_mark = int(input("Оценка по физике от 1 до 5: "))
inf_mark = int(input("Оценка по информатике от 1 до 5: "))

# 1.2 
average = round(sum((math_mark, phy_mark, inf_mark)) / 3, 2)
print(f"Средний балл студента {name}: {average}")

# 1.3
rating = round(math_mark * 0.4 + phy_mark * 0.3 + inf_mark * 0.3, 2)
print(f"Рейтинг студента {name}: {rating}")

# 1.4
if average > rating:
    print(f"Средний балл больше рейтинга на {round((average - rating), 2)}")
elif average < rating:
    print(f"Рейтинга больше среднего балла на {round((rating - average), 2)}")
else:
    print("Средний балл и рейтинг равны")

print("Является ли рейтинг целым: ", rating.is_integer())

# 1.5
average_bool = math_mark >= 4 and phy_mark >= 4 and inf_mark >= 4
print(average_bool)

# 1.6 Не понимаю, как можно сделать рамку, если все надо в одну строку
# А так же не ясно, что такое тип стипендии
print(f"===========\nИмя: {name}\nСр. балл: {average}\nРейтинг: {rating}\n===========")

# 1.7 

is_golden = average_bool and rating >= 4.5 and average >= 4.5
print("Золотой: ", is_golden)




```
