```python
n: int = 0 # <- 1.5
d: float = 0
s: float = 0

price: float = -1
count: int = 0
max: float = 0

print("ДОБРО ПОЖАЛОВАТЬ В       ")
print("МАГАЗИН 'ШЕСТОРОЧКА'     ")

while price != 0:
    try: # 1.1 Ввод данных
        price = float(input("Введите цену товара: "))
        if price == 0: # 1.3
            continue
        count = int(input("Введите колличество: "))

    except ValueError: # 1.2 Проверка корректрости ввода данных
        print("Некорректный ввод данных, попробуйте ещё раз")
        continue
    
    # 1.5 Прибавляем товар
    n += 1 
    s += price * count

    # 1.7 В задании явно не указано, нужно ли учитывать скидку, судя по формуле - нет
    max = price * count if price * count > max else max

    if count > 10: # 1.1 Высчитываем и прибавляем скидку
        d += price * count / 10
    
    print(f"Добавлено {count} товаров по цене {price}")


# 1.6 Увеличиваем скидку относительно всей суммы, если она привышает 5000,
# учитывая скидки на отдельные товары
d += 0 if s - d <= 5000 else (s - d) / 20


# 1.8
print("=========================")
print("          $$$$$          ")
print("         $$              ")
print("         $$$$$$          ")
print("         $$   $$         ")
print("          $$$$$          ")
print("=========================")
print("Позиций:            ", n) # <- 1.5
print("Сумма без скидки:   ", s)
print("Скидка:             ", d)
print("Итого к оплате:     ", s - d) # <- 1.4
print("Самый дорогой товар:", max)
print("=========================")
```
