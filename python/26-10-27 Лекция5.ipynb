```python
class Product:
    def __init__(self, name: str):
        self.name = name

    def __str__(self): # определяет значение при переводе класса в str
        return f"Товар: {self.name}"

prod = Product("Хлебушек")
print(prod)
```

```python
# Реализация дата класса
from dataclasses import dataclass

@dataclass
class NewProduct:
    name: str

    def __str__(self): # все равно можно задавать функции внутри класса
        return f"Товар: {self.name}"


prod2 = NewProduct("Хлебушек2")
print(prod2)
```
