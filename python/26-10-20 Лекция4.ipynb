# Массивы - списки - NumPy

## Списки python

```python
marks = [4, 5, 4, 5, 4]
print(marks[0]) # <- Обращение к элементу по индексу (0)
print(marks[1:4]) # <- срез эл-том
```

```python
text = "10 20 30 40"
l = list(map(int, text.split()))
print(l)
print(sum(l))
```

### Список != числовому массиву

## Numpy

Библиотека для числовых вычислений.
- Базовый объект - ndarray (N-dimensional array)

```python
import numpy as np # <- Импорт numpy

a = np.array([10, 20, 30]) # <- Создание массива
print(a)
print(type(a))
```

```python
import numpy as np

a = [1, 2, 3]
b = np.array([1, 2, 3])
print(a + a)
print(b + b)
```

```python
import numpy as np

a = np.array([[1, 2, 3], # <- Создание двумерного массива (таблицы)
              [4, 5, 6]])

print(a)
```

```python
import numpy as np

print(np.zeros((2, 3), dtype=int))
print(np.ones((2, 2), dtype=int))
print(np.full(1, 2))

print(np.arange(1, 10, 2))
print(np.linspace(0, 1, 5))
```

```python
import numpy as np

a = np.array([[1, 2, 3], 
              [4, 5, 6]])
print(a.ndim)
print(a.shape)
print(a.size)
print(a.dtype)
```

```python
import numpy as np

a = np.array([1, 2, 3], dtype=float)
b = a.astype(int)
print(a)
print(b)
```

```python
import numpy as np

a = np.arange(12)
b = a.reshape(3, 4)
print(a)
print(b)
print(b.shape)
```
