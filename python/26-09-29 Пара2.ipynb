# try except else finally

```python
try:
    number = int(input("Число: "))
    print(100 / number)
except ValueError:
    print("Неверный ввод")
except ZeroDivisionError:
    print("Введен 0")
finally: 
    print("Попытка звершена")

    
```

```python
while True:
    try:
        age = int(input("Возраст: "))
        if not 0 <= age <= 120:
            print("Возраст введен не в диапазоне 0-120")
            continue
    except ValueError:
        print("Введены некорректные данные")

```

# Самостоятельный выброс исключений - `raise`

```python
def set_age(age: int):
    if age < 0:
        raise ValueError("Возраст не может быть отрицательным")
    else:
        return
```

```python
def withdraw(balance, amount):
    if amount < 0:
        raise ValueError("Сумма должна быть положительной")
    if amount > balance:
        raise ValueError("Сумма должна быть меньше или равна балансу")
```

# Собственный класс исключение

```python
class InsufficientFundsError(Exception):
    value: int
    def __init__(self, message, value):
        super().__init__(message)
        self.value = value

def count(i: int) -> int:
    if i < 0:
        raise InsufficientFundsError("Отрицательное значение", i)
    else:
        return i

try:
    print(count(2))
except InsufficientFundsError as e:
    print(f"{e}: {e.value}")
```
